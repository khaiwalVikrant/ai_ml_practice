```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
df = load_iris()

df
```

```python
X, y = load_iris(return_X_y=True)
```

```python
sse = [] #SUM OF SQUARED ERROR
for k in range(1,11):
	km = KMeans(n_clusters=k, random_state=2)
	km.fit(X)
	sse.append(km.inertia_)
```

```python
sns.set_style("whitegrid");
g=sns.lineplot(x=range(1,11), y=sse);

g.set(xlabel ="Number of cluster (k)",
	ylabel = "Sum Squared Error",
	title ='Elbow Method');
```

#### K = 2

```python
# Build models with K = 3
kmeans_2 = KMeans(n_clusters = 2, random_state = 2)
kmeans_2.fit(X)
```

```python
kmeans_2.cluster_centers_
```

```python
pred_2 = kmeans_2.predict(X)
pred_2
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.scatter(X[:,0],X[:,1],c = pred_2, cmap=cm.Accent)
plt.grid(True)
for center in kmeans_2.cluster_centers_:
    center = center[:2]
    plt.scatter(center[0],center[1],marker = '^',c = 'red')
plt.xlabel("petal length (cm)")
plt.ylabel("petal width (cm)")

plt.subplot(1,2,2)
plt.scatter(X[:,2],X[:,3],c = pred_2, cmap=cm.Accent)
plt.grid(True)
for center in kmeans_2.cluster_centers_:
    center = center[2:4]
    plt.scatter(center[0],center[1],marker = '^',c = 'red')
plt.xlabel("sepal length (cm)")
plt.ylabel("sepal width (cm)")
```

#### K = 3

```python
# Build models with K = 3
kmeans = KMeans(n_clusters = 3, random_state = 2)
kmeans.fit(X)
```

```python
kmeans.cluster_centers_
```

```python
pred = kmeans.predict(X)
pred
```
