```python
### Bag of Words
documents = [
    "The cat sat on the mat.",
    "The dog played in the garden.",
    "The sun is shining brightly."
]

# Toeknization
tokenized_documents = [doc.lower().split() for doc in documents]

# Building vocabulary
vocabulary = set()
for doc in tokenized_documents:
  vocabulary.update(doc)

# Counting word frequency
word_frequency = {word: [doc.count(word) for doc in tokenized_documents] for word in vocabulary}

# Creating Document Vectors
document_vactors = []
for doc in tokenized_documents:
  vector = [doc.count(word) for word in vocabulary]
  document_vactors.append(vector)

print(f"Vocabulary: {vocabulary}")
print(f"Word Frequency: {word_frequency}")
print(f"Document Vactors: {document_vactors}")
```

```python
### Count Vectorizer

from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Fit and transform the documents
document_vactors = vectorizer.fit_transform(documents)

# Convert sparse matrix to dense array for redability
dense_document_vectors = document_vactors.toarray()

# Get the vocubalary
vocabulary = vectorizer.get_feature_names_out()

print(f"Vocabulary: {vocabulary}")
print(f"Vectors: {document_vactors}")
```

```python
### TF-IDF

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents
documents = [
    "The cat sat on the mat.",
    "The dog played in the garden.",
    "The sun is shining brightly."
]

# Initialize TfidfVectorizer
vectorizer = TfidfVectorizer()

# Fit and transform the documents
tfidf_document_vectors = vectorizer.fit_transform(documents)

# Convert sparse matrix to Pandas DataFrame
df = pd.DataFrame(tfidf_document_vectors.toarray(), columns=vectorizer.get_feature_names_out())

# Print DataFrame
print("TF-IDF Document Vectors:", df)
print(df)
```

```python
### N-grams

documents = [
    "The cat sat on the mat.",
    "The dog played in the garden.",
    "The sun is shining brightly.",
    "The cat played in the garden."
]

# Initialize CountVectorizer for bigrams
vectorizer = CountVectorizer(ngram_range=(3, 3))  # ngram_range=(2, 2) for bigrams

# Fit and transform the documents
ngram_vectors = vectorizer.fit_transform(documents)

# Convert sparse matrix to Pandas DataFrame
df = pd.DataFrame(ngram_vectors.toarray(), columns=vectorizer.get_feature_names_out())

# Print DataFrame
print("Bigram Document Vectors:")
print(df)
```
