```python
def null_percentage_data(data):
    df = data.copy()
    missing = pd.DataFrame(100*df.isnull().sum()/len(df.index))
    missing.rename(columns={0:'Null_Percentage'}, inplace=True)
    missing = pd.DataFrame(missing.Null_Percentage).sort_values(by='Null_Percentage',ascending=False)
    return missing
```

```python
churn_data = churn_data.rename(columns= lambda x: x.strip(), inplace=False)
```

```python
churn_data = churn_data.drop_duplicates()
```

In churn prediction, we are given that there are three phases of customer lifecycle :


1.   The **good** phase [Month 6 & 7]: First two months
2.   The **action** phase [Month 8]: Third month
3.   The **churn** phase [Month 9]: Fourth Month


```python
# Filter both reacharge amount and recharge data
recharge_data_cols = churn_data.columns[churn_data.columns.str.contains("rech_amt|rech_data", regex=True)]
print(recharge_data_cols)
```

Observation: The above mentioned columns correspond to **rech_amt** and **rech_data** features across months 6, 7 and 8 respectively:
- Total recharge amount (calling)
- Maximum recharge amount (calling)
- Last day recharge amount
- Last date of Recharging the data
- Total recharge for data
- Maximum recharge for data
- Average recharge amount for data

```python
recharge_and_data_df = churn_data[recharge_data_cols]
# recharge_and_data_df.columns
null_percentage_data(recharge_and_data_df)
```

```python
recharge_and_data_df.describe()
```

```python
##Imputation of 'av_rech_amt_data_* and total_rech_data_*' column NaNs with 0 (for all the months 6, 7, 8 and 9 respectively)

churn_data['av_rech_amt_data_6']= churn_data['av_rech_amt_data_6'].fillna(0)
churn_data['av_rech_amt_data_7']= churn_data['av_rech_amt_data_7'].fillna(0)
churn_data['av_rech_amt_data_8']= churn_data['av_rech_amt_data_8'].fillna(0)
churn_data['av_rech_amt_data_9']= churn_data['av_rech_amt_data_9'].fillna(0)

churn_data['total_rech_data_6']= churn_data['total_rech_data_6'].fillna(0)
churn_data['total_rech_data_7']= churn_data['total_rech_data_7'].fillna(0)
churn_data['total_rech_data_8']= churn_data['total_rech_data_8'].fillna(0)
churn_data['total_rech_data_9']= churn_data['total_rech_data_9'].fillna(0)
```

```python
##Deriving features to subset the high-value customers.

##Calculating total data recharge amt for all the months 6, 7, 8 and 9 respectively.
churn_data['total_rech_data_amt_6'] = (churn_data['av_rech_amt_data_6'] * churn_data['total_rech_data_6'])
churn_data['total_rech_data_amt_7'] = (churn_data['av_rech_amt_data_7'] * churn_data['total_rech_data_7'])
churn_data['total_rech_data_amt_8'] = (churn_data['av_rech_amt_data_8'] * churn_data['total_rech_data_8'])
churn_data['total_rech_data_amt_9'] = (churn_data['av_rech_amt_data_9'] * churn_data['total_rech_data_9'])
```

```python
##Calculating the total average recharge for good phase: average of (calling recharge costs + data recharge costs) for month (6)
##and (7). Note: Good phase: month (6) and (7)

churn_data['avg_total_rech_plus_data_amt_6_7'] = (churn_data['total_rech_data_amt_6'] + churn_data['total_rech_amt_6']+
                                                  churn_data['total_rech_data_amt_7'] + churn_data['total_rech_amt_7'])/2
```

**Filter High-Value customers**

Criteria for filtering High-Value Customers:
High-valued customers are those customers who have recharged with an amount more than or equal to X, where X is the **70th percentile** of the average recharge amount in the first two months (the good phase)

Creating a variable **avg_total_rech_plus_data_amt_6_7_70per** to filter high-value customers.

```python
avg_total_rech_plus_data_amt_6_7_70per =  churn_data['avg_total_rech_plus_data_amt_6_7'].quantile(0.7)
print(avg_total_rech_plus_data_amt_6_7_70per)
```

```python
highvalue_customer = churn_data.loc[(churn_data.avg_total_rech_plus_data_amt_6_7 > avg_total_rech_plus_data_amt_6_7_70per)]
highvalue_customer.shape
```

```python
#Tagging churners (churn=1, else 0) based on the fourth month(i.e. month of September(9)) and removing attributes corresponding to the churn phase.

churn_phase_cols = ['total_og_mou_9','total_ic_mou_9', 'vol_2g_mb_9', 'vol_3g_mb_9']

churn_phase_cols_df = churn_data[churn_phase_cols]
null_percentage_data(churn_phase_cols_df)
```

```python
highvalue_customer['total_usage_col_9'] = (churn_data['total_og_mou_9'] + churn_data['total_ic_mou_9'] +
                                   churn_data['vol_2g_mb_9'] + churn_data['vol_3g_mb_9'])
```

```python
highvalue_customer['churn'] = np.where(highvalue_customer['total_usage_col_9']== 0, 1, 0)
```

```python
print("Churn data")
highvalue_customer['churn'].value_counts()
```

```python
print("Churner (%)")
round(100*highvalue_customer['churn'].value_counts()/len(highvalue_customer.index),3)
```

```python
#Churn Rate
churn_rate = round(100*highvalue_customer['churn'].sum()/len(highvalue_customer.index),3)
print('Churn Rate(%):',churn_rate)
print("High-Valued Customers Dataset shape: ",highvalue_customer.shape)
```
